```python
import torch

x = [1,2,3,4,5]   # fill in: numbers 1 to 5
y = [2,4,6,8,10]   # fill in: their doubles

x_tensor = torch.tensor(x, dtype=torch.float32)
y_tensor = torch.tensor(y, dtype=torch.float32)

print(x_tensor)
print(y_tensor)
```

```python
x_tensor.shape
```

```python
x_tensor = x_tensor.reshape(5, -1)
y_tensor = y_tensor.reshape(5, -1)
```

```python
import torch.nn as nn
```

```python
class RegressionModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer = nn.Linear(1, 1)

    def forward(self, x):
        return self.layer(x)
```

```python
model = RegressionModel()
```

```python
print(model)
```

```python
for name, param in model.named_parameters():
    print(name, param.data)
```

```python
with torch.no_grad():
    predictions = model(x_tensor)
    print(predictions)
```

```python
-0.9930*1 -0.3444
```

```python
criterion = nn.MSELoss()
loss = criterion(predictions, y_tensor)
```

```python
print(loss)
```

```python
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)
```

```python
print(optimizer)
```

```python
epochs = 1000

for i in range(epochs):
    # make predictions
    predictions = model(x_tensor)
    # compute loss
    loss = criterion(predictions, y_tensor)
    # zero gradients
    optimizer.zero_grad()
    # backward pass
    loss.backward()
    # update weights
    optimizer.step()
    if i%100 == 0:
        print("Epoch: {} | Loss: {}".format(i, loss.item()))
```
