```python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import root_mean_squared_error, mean_absolute_percentage_error, r2_score
from sklearn.linear_model import LinearRegression, LassoCV, RidgeCV, ElasticNetCV, Lasso
from sklearn.preprocessing import StandardScaler
import numpy as np
```

```python
df = pd.read_csv("./nyc_taxi_preprocessed.csv")
```

```python
df.head()
```

```python
X = df.drop("trip_duration", axis=1)
y = np.log1p(df["trip_duration"])
```

```python
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.3)
```

### Basic Decision Tree Regressor Model

```python
DT_reg_model = DecisionTreeRegressor(random_state=42)
```

```python
DT_reg_model.fit(x_train, y_train)
```

```python
y_pred = DT_reg_model.predict(x_test)
```

```python
root_mean_squared_error(y_test, y_pred), mean_absolute_percentage_error(y_test, y_pred), r2_score(y_test, y_pred)
```

```python
root_mean_squared_error(y_test, y_pred), mean_absolute_percentage_error(y_test, y_pred), r2_score(y_test, y_pred)
```

### Updating the DT model hyperparameters

```python
# List of max_depth values 
max_depth_list = list(range(25,3,-3))
max_depth_list
```

```python
train_score = {}
test_score = {}
for d in max_depth_list:
    dt_model = DecisionTreeRegressor(max_depth=d, random_state=42)
    dt_model.fit(x_train, y_train)
    train_score[d] = mean_absolute_percentage_error(y_train, dt_model.predict(x_train))
    test_score[d] = mean_absolute_percentage_error(y_test, dt_model.predict(x_test))
```

```python
for d in max_depth_list:
    print("Max Dept: {0} | Train score: {1} | Test score: {2}".format(d, train_score[d], test_score[d]))
```

#### Comparing the output with the mean

```python
# Dictionary to store the train and test r square scores

train_scores = {}
test_scores = {}

# Fill the values in the below list and assign value to the variable!
min_sample_leaf_list = list(range(5, 15, 2))
depth = 16

# Loop through min_sample_leaf values and train the models
for min_sample_leaf in min_sample_leaf_list:
    
    # Initialize the Decision Tree model with the current min_samples_leaf value
    DT_model = DecisionTreeRegressor(min_samples_leaf=min_sample_leaf, max_depth=depth, random_state=42)
    
    # Train the model
    DT_model.fit(x_train, y_train)
    
    # Make predictions on the train dataset
    y_train_pred = DT_model.predict(x_train)

    # Make predictions on the test dataset
    y_test_pred = DT_model.predict(x_test)
    
    # Store the train r square score in the dictionary with the min_sample_leaf as the key
    train_scores[min_sample_leaf] = mean_absolute_percentage_error (y_train, y_train_pred)
    
    # Store the test r square score in the dictionary with the min_sample_leaf as the key
    test_scores[min_sample_leaf] = mean_absolute_percentage_error (y_test, y_test_pred)
```

```python
# Print the train and test r square scores for each model
for min_sample_leaf in min_sample_leaf_list:
    print(f"min_sample_leaf = {min_sample_leaf}|\
    Train Score = {train_scores[min_sample_leaf]:.3f} |\
    Test Score = {test_scores[min_sample_leaf]:.3f}")
    print('_'*65)
```

### Applying regularization

```python
# create an array with 20 numbers equllay spaced between 0 to 10
alphas = np.linspace(0, 10, 20)
alphas
```

```python
# Initialize LassoCV to find the best alpha
lasso_cv = LassoCV(alphas= alphas,
                   cv=10, 
                   max_iter=100000)
```

```python
lasso_cv.fit(x_train, y_train)
```

```python
# Find the best alpha value
best_alpha = lasso_cv.alpha_
best_alpha
```

```python
# Compare the coefficients before and after L2 regularization
coefficients_after_l1 = pd.DataFrame({
    'Feature': x_train.columns,
    'After L1 Regularization': lasso_cv.coef_
})
coefficients_after_l1
```

```python
mean_absolute_percentage_error(y_test, y_pred)
```

```python
# create an array with 20 numbers equllay spaced between 0 to 10
alphas = np.linspace(0, 10, 20)
alphas
```

```python
# Initialize ElasticNetCV to find the best alpha and l1_ratio
elastic_net_cv = ElasticNetCV(alphas=alphas, cv=10, max_iter=10000, l1_ratio=[.1, .5, .7, .9, .95, .99, 1])
elastic_net_cv.fit(x_train, y_train)
```

```python
# retrieve the best alpha and l1_ratio
best_alpha_enet = elastic_net_cv.alpha_
best_l1_ratio = elastic_net_cv.l1_ratio_

# print the l1_ratio and best_alpha value
(best_l1_ratio, best_alpha_enet)
```

```python
# R-squared scores for the Ridge model
r2_train_enet = elastic_net_cv.score(x_train, y_train)
r2_test_enet = elastic_net_cv.score(x_test, y_test)

r2_train_enet, r2_test_enet
```
